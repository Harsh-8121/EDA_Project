```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

df = pd.read_csv("/content/WineQT.csv")

print("Dataset Shape:")
print(df.shape)

print("\nFirst 3 Rows:")
print(df.head(3))

print("\nLast 3 Rows:")
print(df.tail(3))
```

```python
print("Data Types of Each Column:")
print(df.dtypes)
```

```python
print("Descriptive Statistics:")
print(df.describe())
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nDoes Each Column Contain Missing Values?")
print(df.isnull().any())
```

```python
mean_quality = df["quality"].mean()
median_quality = df["quality"].median()
mode_quality = df["quality"].mode()

print("Mean Quality:", mean_quality)
print("Median Quality:", median_quality)
print("Mode Quality:", mode_quality.values)
```

```python
std_quality = df["quality"].std()
variance_quality = df["quality"].var()

print("Standard Deviation of Quality:", std_quality)
print("Variance of Quality:", variance_quality)
```

```python
plt.figure(figsize=(10, 5))

df["quality"].value_counts().sort_index().plot(kind="bar")

plt.title("Distribution of Wine Quality")
plt.xlabel("Quality Score")
plt.ylabel("Number of Wines")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(df["alcohol"], kde=True)

plt.title("Distribution of Alcohol")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(x=df["alcohol"])

plt.title("Boxplot of Alcohol")
plt.xlabel("Alcohol")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(df["alcohol"], df["quality"])

plt.title("Scatter Plot of Alcohol vs Wine Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(x="quality", y="alcohol", data=df)

plt.title("Alcohol Distribution Across Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Alcohol")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(df["volatile acidity"], df["quality"])

plt.title("Volatile Acidity vs Wine Quality")
plt.xlabel("Volatile Acidity")
plt.ylabel("Quality")

plt.show()
```

```python
wine_features = [
    "fixed acidity",
    "volatile acidity",
    "citric acid",
    "residual sugar",
    "chlorides",
    "free sulfur dioxide",
    "total sulfur dioxide",
    "density",
    "pH",
    "sulphates",
    "alcohol",
    "quality"
]

wine_data = df[wine_features]

print("Selected Variables:")
print(wine_data.head())
```

```python
sns.pairplot(
    df[
        [
            "alcohol",
            "volatile acidity",
            "citric acid",
            "sulphates",
            "quality"
        ]
    ],
    kind="reg"
)

plt.show()
```

```python
from scipy import stats

corr = stats.pearsonr(
    df["alcohol"],
    df["quality"]
)

print("Pearson Correlation:")
print("p-value:", corr[1])
print("Correlation:", corr[0])
```

```python
corr = stats.pearsonr(
    df["volatile acidity"],
    df["quality"]
)

print("Pearson Correlation:")
print("p-value:", corr[1])
print("Correlation:", corr[0])
```

```python
correlation = df.corr(
    method="pearson",
    numeric_only=True
)

print("Pearson Correlation Matrix:")
print(correlation)
```

```python
quality_correlation = correlation["quality"].sort_values(
    ascending=False
)

print("Correlation of Features with Wine Quality:")
print(quality_correlation)
```

```python
plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Pearson Correlation Heatmap of Wine Dataset")

plt.show()
```

```python
quality_corr = correlation[["quality"]].sort_values(
    by="quality",
    ascending=False
)

plt.figure(figsize=(6, 10))

sns.heatmap(
    quality_corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation of Wine Features with Quality")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="alcohol",
    y="sulphates",
    hue="quality"
)

plt.title("Alcohol vs Sulphates by Wine Quality")
plt.xlabel("Alcohol")
plt.ylabel("Sulphates")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="fixed acidity",
    y="volatile acidity",
    hue="quality"
)

plt.title("Fixed Acidity vs Volatile Acidity by Wine Quality")
plt.xlabel("Fixed Acidity")
plt.ylabel("Volatile Acidity")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="fixed acidity",
    y="volatile acidity",
    hue="quality"
)

plt.title("Fixed Acidity vs Volatile Acidity by Wine Quality")
plt.xlabel("Fixed Acidity")
plt.ylabel("Volatile Acidity")

plt.show()
```

```python
print("Strongest Positive Correlations with Quality:")
print(
    quality_correlation[
        quality_correlation > 0
    ].sort_values(ascending=False)
)

print("\nStrongest Negative Correlations with Quality:")
print(
    quality_correlation[
        quality_correlation < 0
    ].sort_values()
)
```
