```python
import numpy as np
import pandas as pd

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.impute import SimpleImputer

# Creating student dataset
data = pd.DataFrame({
    "Study_Hours": [4, np.nan, 6, 8, 3, 5],
    "Exam_Score": [65, 72, np.nan, 88, 55, 70],
    "Attendance": [80, 85, 90, np.nan, 75, 82],
    "Department": ["CSE", "ECE", "CSE", np.nan, "MECH", "ECE"]
})

print("Original Data:")
print(data)
```

```python
# Median Imputation
median_imputer = SimpleImputer(strategy="median")

data["Exam_Score_Median"] = median_imputer.fit_transform(
    data[["Exam_Score"]]
)

print("After Median Imputation:")
print(data[["Exam_Score", "Exam_Score_Median"]])
```

```python
# Median Imputation
median_imputer = SimpleImputer(strategy="median")

data["Exam_Score_Median"] = median_imputer.fit_transform(
    data[["Exam_Score"]]
)

print("After Median Imputation:")
print(data[["Exam_Score", "Exam_Score_Median"]])
```

```python
# Mode Imputation

mode_imputer = SimpleImputer(strategy="most_frequent")

data["Department_Mode"] = mode_imputer.fit_transform(
    data[["Department"]]
).ravel()

print("After Mode Imputation:")
print(data[["Department", "Department_Mode"]])
```

```python
# MICE Imputation
numerical_columns = [
    "Study_Hours",
    "Exam_Score",
    "Attendance"
]

mice_imputer = IterativeImputer(
    max_iter=10,
    random_state=42
)

mice_data = data[numerical_columns].copy()

mice_data = pd.DataFrame(
    mice_imputer.fit_transform(mice_data),
    columns=numerical_columns
)

data["Study_Hours_MICE"] = mice_data["Study_Hours"]
data["Exam_Score_MICE"] = mice_data["Exam_Score"]
data["Attendance_MICE"] = mice_data["Attendance"]

print("After MICE Imputation:")
print(data[[
    "Study_Hours_MICE",
    "Exam_Score_MICE",
    "Attendance_MICE"
]])
```

```python
print("Final Dataset:")
print(data)
```
