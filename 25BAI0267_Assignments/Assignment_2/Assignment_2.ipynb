```python
import pandas as pd
df1SE = pd.DataFrame({
    'StudentID': [101, 103, 105, 107, 109, 111, 113, 115, 117, 119, 121],
    'ScoreSE': [65, 72, 58, 81, 69, 75, 88, 54, 79, 91, 67]
})

df2SE = pd.DataFrame({
    'StudentID': [102, 104, 106, 108, 110, 112, 114, 116, 118, 120, 122],
    'ScoreSE': [73, 84, 62, 77, 85, 68, 90, 74, 59, 83, 71]
})

df1ML = pd.DataFrame({
    'StudentID': [101, 103, 105, 107, 109, 111, 113, 115, 117, 119, 121],
    'ScoreML': [70, 80, 66, 89, 74, 92, 78, 63, 81, 87, 72]
})

df2ML = pd.DataFrame({
    'StudentID': [102, 104, 106, 108, 110, 112, 114, 116, 118, 120, 122],
    'ScoreML': [76, 85, 69, 82, 90, 73, 88, 65, 79, 91, 75]
})
```

```python
# Concatenation

dfSE = pd.concat([df1SE, df2SE], ignore_index=True)
dfML = pd.concat([df1ML, df2ML], ignore_index=True)
```

```python
# Column-wise Concatenation

df = pd.concat([dfML, dfSE], axis=1)
df
```

```python
# Inner Join

df = dfSE.merge(dfML, how='inner')
df
```

```python
# Inner Join

df = dfSE.merge(dfML, how='inner')
df
```

```python
# Right Join

df = dfSE.merge(dfML, how='right')
df
```
