```python
import numpy as np
import pandas as pd

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import SimpleImputer, IterativeImputer

student_data = pd.DataFrame({
    "Marks": [82, np.nan, 74, 91, 67, np.nan],
    "Study_Hours": [5, 7, np.nan, 8, 4, 6],
    "Section": ["A", "B", np.nan, "A", "C", "B"]
})

print("Original Data")
student_data
```

```python
mean_fill = SimpleImputer(strategy="mean")

student_data["Marks_Mean"] = mean_fill.fit_transform(student_data[["Marks"]])

print("After Mean Imputation")
student_data
```

```python
median_fill = SimpleImputer(strategy="median")

student_data["Hours_Median"] = median_fill.fit_transform(student_data[["Study_Hours"]])

print("After Median Imputation")
student_data
```

```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer

df = pd.DataFrame(
    {
        "Height": [165, np.nan, 172, 180, 168],
        "Weight": [60, 72, np.nan, 85, 65],
        "Department": ["IT", "CSE", "IT", np.nan, "ECE"],
    }
)

mode_imputer = SimpleImputer(strategy="most_frequent")
df["Department_Mode"] = mode_imputer.fit_transform(df[["Department"]]).ravel()

print(df)
```

```python
numeric_cols = ["Marks", "Study_Hours"]

mice = IterativeImputer(
    max_iter=10,
    random_state=10
)

temp = student_data[numeric_cols].copy()

temp.iloc[:, :] = mice.fit_transform(temp)

student_data["Marks_MICE"] = temp["Marks"]
student_data["Hours_MICE"] = temp["Study_Hours"]

print("After MICE Imputation")
student_data
```
