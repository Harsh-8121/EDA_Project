```python
correlation = df[
    [
        "SepalLengthCm",
        "SepalWidthCm",
        "PetalLengthCm",
        "PetalWidthCm"
    ]
].corr(method="pearson")

print("Pearson Correlation Matrix:")
print(correlation)
```
