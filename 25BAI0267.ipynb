```python
# Checking missing values

import numpy as np
import pandas as pd

data = pd.read_csv("employee_salary.csv")

total = data.isnull().sum().sort_values(ascending=False)
percent = (data.isnull().sum()/data.isnull().count()).sort_values(ascending=False)
missing_data = pd.concat([total, percent], axis=1, keys=["Total","Percent"])

missing_data
```

```python
import numpy as np
import pandas as pd

data = pd.read_csv("employee_salary.csv")

# Converting Salary from '12L' to 12 that is string to int
data["Salary"] = data["Salary"].str.replace("L", "", regex=False).astype(float)

from sklearn.preprocessing import StandardScaler

salary_scaled = StandardScaler().fit_transform(data[['Salary']])

low_range = salary_scaled[salary_scaled[:,0].argsort()][:5]

high_range = salary_scaled[salary_scaled[:,0].argsort()][-5:]

print("Outer Range (Low)")
print(low_range)

print()

print("Outer Range (High)")
print(high_range)
```
